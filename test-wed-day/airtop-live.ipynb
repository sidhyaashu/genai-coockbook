```python
import os 
from dotenv import load_dotenv

load_dotenv()

api_key = os.getenv("AIRTOP")
```

```python
from airtop import Airtop

client = Airtop(api_key=api_key)
```

```python
session = client.sessions.create()

window = client.windows.create(session.data.id, url="https://www.airtop.ai")
```

```python
window_info = client.windows.get_window_info(session.data.id, window.data.window_id)
live_view_url = window_info.data.live_view_url
```

```python
print(window.data.window_id)

print(live_view_url)
```
